```python
import pandas as pd
df=pd.read_csv('/content/Global_Superstore2.csv.zip',encoding='latin1')
df.head()
```

```python
df['Postal Code']=df['Postal Code'].fillna('Not Available')
```

```python
df.dtypes
```

```python
df['Order Date']=pd.to_datetime(df['Order Date'], format='%d-%m-%Y')
df['Ship Date']=pd.to_datetime(df['Ship Date'], format='%d-%m-%Y')
```

```python
df.dtypes
```

```python
df['Oder Year']=df['Order Date'].dt.year
df['Order Month']=df['Order Date'].dt.month
```

```python
df.to_csv('Global_Superstore_Cleaned.csv', index=False)
```
